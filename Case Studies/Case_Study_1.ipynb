## Tokenization

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize

sent_tokenize(cleaned_text)
```

```python
from nltk.tokenize import word_tokenize
word_tokens=word_tokenize(cleaned_text)
word_tokens
```

## Stopwors Removal

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
stop_words=stopwords.words('english')
filtered=[word for word in word_tokens if word not in stop_words]
print(filtered)
```

## stemming

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
for word in filtered:
  print(word,"->",stemmer.stem(word))
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()

tokens=[]
words=[]
for word in filtered:
  tokens.append(lemmatizer.lemmatize(word,pos='v'))
for word in tokens:
  words.append(lemmatizer.lemmatize(word,pos='a'))
words
```

## Vectorization

### OneHotEncoding


```python
import numpy as np
words = np.array(words).reshape(-1, 1)
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
encoded=encoder.fit_transform(words)
encoded.toarray()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
words = words.ravel()
vectorizer=CountVectorizer()
vectorized=vectorizer.fit_transform(words)
print(vectorizer.get_feature_names_out())
sparse_matrix=vectorized.toarray()
print(sparse_matrix)
print(vectorizer.vocabulary_)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(words)

print(vectorizer.get_feature_names_out())

print(tfidf.toarray())
```
